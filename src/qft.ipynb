```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
from qiskit import quantum_info as qi
from qiskit.visualization import array_to_latex, plot_bloch_multivector, plot_histogram
from qiskit_aer import Aer

from qiskit.synthesis.qft import synth_qft_full
```

## Quantum Fourier Transform

```python
qc = QuantumCircuit(1)   # initialize a quantum circuit
qc.initialize("0", qc.qubits)   # initialize a state vector 

qc.compose(synth_qft_full(1), qubits = qc.qubits, inplace = True)
```

```python
qc.draw("mpl")
```

```python
psi_f = qi.Statevector.from_circuit(qc)
array_to_latex(psi_f)
```

#### Quantum Fourier Transform for different 2 cubit bases

$$
\lvert 00 \rangle = \begin{pmatrix}1\\0\\0\\0\end{pmatrix}, \quad
\lvert 01 \rangle = \begin{pmatrix}0\\1\\0\\0\end{pmatrix}, \quad
\lvert 10 \rangle = \begin{pmatrix}0\\0\\1\\0\end{pmatrix}, \quad
\lvert 11 \rangle = \begin{pmatrix}0\\0\\0\\1\end{pmatrix}
$$

```python
qc = QuantumCircuit(2)   # initialize a quantum circuit
qc.initialize("00", qc.qubits)   # initialize a state vector 

qc.compose(synth_qft_full(2), qubits = qc.qubits, inplace = True)
```

```python
qc.draw("mpl")
```

Final state after transformation (Quantum Circuit)

```python
psi_f = qi.Statevector.from_circuit(qc)
array_to_latex(psi_f)
```
